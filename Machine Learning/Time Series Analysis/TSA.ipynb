### Time Series Analysis

Time series analysis is a statistical technique used to analyze sequentially ordered data points collected over time. It focuses on identifying patterns, trends, seasonality, and other temporal characteristics to understand underlying processes and make predictions. Time series analysis is widely used across fields including finance, economics, environmental science, signal processing, and IoT applications.

The process typically involves decomposing a time series into its constituent components (trend, seasonality, cyclical, and irregular components), then applying appropriate statistical methods to model these components. Time series models can be used for forecasting future values, detecting anomalies, understanding temporal relationships, and inferring causal connections between variables.

Time series analysis differs from conventional data analysis in its explicit recognition of the importance of data ordering and temporal dependence between observations. This requires specialized techniques that account for autocorrelation, stationarity, and other time-specific properties.

#### Pros
- **Captures temporal patterns** that cross-sectional analysis would miss
- **Enables forecasting** of future values based on historical patterns
- **Provides insights into system dynamics** and how variables evolve over time
- **Identifies seasonal patterns** and cyclical behaviors
- **Detects anomalies** and structural changes in temporal data
- **Quantifies uncertainty** in temporal predictions with confidence intervals
- **Supports causal inference** through techniques like Granger causality

#### Cons
- **Requires sufficient historical data** to identify patterns reliably
- **Assumes patterns will continue** in the future (stationarity assumption)
- **Sensitive to outliers** and missing values
- **Complex to model** multiple interacting factors simultaneously
- **Computationally intensive** for large datasets or complex models
- **Risk of overfitting** when using highly parameterized models
- **Challenging to validate** as future data isn't available during model development

#### Key Components & Concepts

##### Time Series Decomposition

Decomposition breaks down a time series into meaningful components that are easier to analyze individually.

| Component | Description | Visual Characteristics | Analysis Approach |
|-----------|-------------|------------------------|-------------------|
| Trend | Long-term progression of the series | Smooth line showing direction | Moving averages, regression |
| Seasonality | Regular, calendar-related patterns | Repeating patterns of fixed length | Seasonal differencing, seasonal indices |
| Cyclical | Irregular fluctuations, longer than seasonality | Waves without fixed frequency | Filtering, spectral analysis |
| Irregular | Random variation (noise) | Unpredictable fluctuations | Statistical distribution analysis |
| Level | Baseline value of the series | Horizontal line (if stationary) | Mean calculation after detrending |

##### Stationarity and Transformations

Most time series methods require stationarity - consistent statistical properties over time.

| Transformation | Purpose | Implementation | When to Use |
|----------------|---------|----------------|------------|
| Differencing | Remove trend and make series stationary | Calculate differences between consecutive observations | Non-stationary data with clear trends |
| Logarithmic | Stabilize variance, handle exponential growth | Take natural log of values | Data with increasing variance or exponential growth |
| Box-Cox | Comprehensive family of power transformations | Apply power transformation with optimized parameter | Non-normal residuals, heteroscedasticity |
| Seasonal Differencing | Remove seasonal patterns | Calculate differences between observations at seasonal intervals | Strong seasonal patterns |
| Moving Average | Smooth fluctuations | Calculate average over sliding window | Noisy data, identifying trends |
| Detrending | Remove trend component | Subtract fitted trend from original series | Isolating non-trend components |

##### Time Series Modeling Approaches

Various modeling approaches exist for different types of time series and analysis objectives.

| Model Type | Description | Best For | Limitations |
|------------|-------------|----------|------------|
| ARIMA | AutoRegressive Integrated Moving Average | Univariate forecasting, stationary series | Assumes linearity, limited seasonality handling |
| SARIMA | Seasonal ARIMA | Seasonal data with clear patterns | More parameters, needs longer history |
| Exponential Smoothing | Weighted averages with exponentially decreasing weights | Short-term forecasting, trending data | Less theoretical foundation than ARIMA |
| VAR | Vector AutoRegression | Multivariate time series, interdependencies | Requires stationarity, vulnerable to overfitting |
| GARCH | Generalized AutoRegressive Conditional Heteroskedasticity | Volatility modeling, financial series | Only models variance, not level |
| Prophet | Decomposable model with trend, seasonality, holidays | Business forecasting, multiple seasonalities | Less flexible than fully parametric models |
| State Space Models | Dynamic systems with hidden states | Complex systems, filtering and smoothing | More complex to implement and interpret |
| Deep Learning | RNNs, LSTMs, TCNs for sequential data | Complex nonlinear patterns, large datasets | Requires substantial data, black-box nature |

##### Forecast Accuracy Metrics

Various metrics are used to evaluate time series forecasting performance.

| Metric | Formula | Interpretation | Best For |
|--------|---------|----------------|----------|
| MAE | Mean Absolute Error | Average absolute difference between predictions and actuals | General-purpose, intuitive |
| RMSE | Root Mean Square Error | Square root of average squared errors | Penalizes large errors more heavily |
| MAPE | Mean Absolute Percentage Error | Average percentage error | Comparing across different scales |
| SMAPE | Symmetric MAPE | Symmetric version of MAPE | Handles zero or near-zero values better |
| MASE | Mean Absolute Scaled Error | MAE relative to naïve forecast | Comparing across different series |
| Theil's U | Ratio of forecasting error to naïve method error | Values < 1 indicate model better than naïve | Assessing predictive value added |
| ACF of Residuals | Autocorrelation Function of model residuals | Should show no pattern if model is adequate | Diagnostic checking |

#### Common Time Series Analysis Workflows

##### Univariate Forecasting Workflow
1. **Data Preparation:**
   - Check for missing values and outliers
   - Transform to appropriate frequency (daily, weekly, etc.)
   - Create training and test sets (hold out recent data)
   - Visualize the data and identify obvious patterns

2. **Time Series Decomposition:**
   - Decompose into trend, seasonal, cyclical, and irregular components
   - Identify multiple seasonalities if present (daily, weekly, yearly)
   - Examine components for insights about the underlying process

3. **Stationarity Testing and Transformation:**
   - Perform ADF and KPSS tests for stationarity
   - Apply appropriate transformations (differencing, log, Box-Cox)
   - Re-test stationarity after transformations

4. **Model Identification:**
   - Examine ACF/PACF plots to identify potential model orders
   - Consider several candidate models (ARIMA, ETS, etc.)
   - Select initial parameters based on data characteristics

5. **Model Estimation and Selection:**
   - Fit multiple models with different parameters
   - Compare models using information criteria (AIC, BIC)
   - Perform residual diagnostics to check model adequacy
   - Implement time series cross-validation if data permits

6. **Forecasting and Evaluation:**
   - Generate forecasts for test period
   - Calculate forecast accuracy metrics (MAE, RMSE, MAPE)
   - Compare with benchmark models (naïve, seasonal naïve)
   - Create prediction intervals to quantify uncertainty
   - Visualize forecasts alongside actual values

7. **Model Refinement:**
   - Adjust model based on diagnostics and accuracy
   - Consider ensemble approaches or combining forecasts
   - Implement automated parameter re-estimation for production

##### Multivariate Analysis Workflow
1. **Data Preparation:**
   - Align multiple time series to common frequency
   - Handle missing values across series
   - Test for cointegration if dealing with economic/financial data
   - Create train/test splits preserving temporal order

2. **Exploratory Analysis:**
   - Calculate cross-correlations between series
   - Identify lead/lag relationships
   - Test for Granger causality between variables
   - Create scatter plots at different lags to visualize relationships

3. **Feature Engineering:**
   - Create lagged features from all series
   - Generate moving averages and other derived features
   - Incorporate external regressors (e.g., weather, calendar events)
   - Consider dimensionality reduction for many variables

4. **Modeling Strategy Selection:**
   - Choose between VAR, VARMAX, regression with ARIMA errors, etc.
   - For many variables, consider factor models or shrinkage methods
   - Implement machine learning approaches for complex relationships

5. **Model Estimation and Validation:**
   - Fit models with appropriate lag structures
   - Perform model diagnostics (residual analysis, stability testing)
   - Validate with out-of-sample testing
   - Implement rolling window validation for time-varying relationships

6. **Forecasting and Scenario Analysis:**
   - Generate point forecasts and prediction intervals
   - Perform impulse response analysis for shock propagation
   - Conduct scenario analysis with different paths for exogenous variables
   - Compare conditional vs. unconditional forecasts

7. **Deployment and Monitoring:**
   - Implement regular model updates as new data arrives
   - Monitor forecast errors for potential model degradation
   - Set up alerts for structural breaks or regime changes
   - Create visualization dashboard for ongoing assessment

#### Tips and Tricks for Effective Time Series Analysis

| Tip | Why It Helps | Implementation |
|-----|--------------|----------------|
| Handle calendar effects | Accounts for holidays, trading days, etc. | Include dummy variables or use specialized packages |
| Detect and handle outliers | Prevents distortion of model estimates | Robust methods, intervention analysis |
| Pre-process at appropriate frequency | Matches analysis to decision timescale | Aggregate or disaggregate with proper methods |
| Test multiple decomposition methods | Different methods reveal different patterns | Compare STL, X-12-ARIMA, and model-based approaches |
| Use dynamic transformations | Adapts to changing data characteristics | Rolling window normalizations, adaptive transforms |
| Incorporate domain knowledge | Improves model specification and interpretation | Add known constraints, seasonal patterns, events |
| Ensemble multiple methods | Reduces risk of method-specific errors | Weighted combinations of diverse models |
| Monitor forecast performance | Detects when models need updating | Track errors over time, implement automated retraining |
| Validate with multiple metrics | Different metrics reveal different weaknesses | Create scorecards with diverse error measures |
| Consider prediction intervals | Quantifies forecast uncertainty | Bootstrap, parametric, or Bayesian approaches |


```python
%pip install --quiet tensorflow pandas scikit-learn matplotlib seaborn statsmodels

import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.metrics import mean_squared_error, mean_absolute_error
from statsmodels.tsa.arima.model import ARIMA
```

```python
df = pd.read_csv('combined_weather.csv',index_col=0)
df.head(2)
```

```python
plt.figure(figsize=(12, 14))
correlation_matrix = df.corr(numeric_only=True)
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
```

```python
# Remove multiple columns
df = df.drop(['preciptype', 'sunrise', 'sunset','tempmax', 'tempmin','conditions', 'description', 'icon', 'stations','snow','snowdepth'], axis=1)


correlation_threshold = 0.9

# Calculate correlation matrix
correlation_matrix = df.corr(numeric_only=True).abs()

# Upper triangle to remove duplicate correlations (upper triangular part)
upper_triangle = correlation_matrix.where(np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool))

# Find columns with correlation greater than the threshold
to_drop = [column for column in upper_triangle.columns if any(upper_triangle[column] > correlation_threshold)]

# Drop the columns with high correlation
df.drop(columns=to_drop, inplace=True)

# Output the dropped columns
print(f"Dropped columns: {to_drop}")
```

```python
numeric_df = df.select_dtypes(include=[np.number])
target_corr = numeric_df.corr()
plt.figure(figsize=(12, 14))
sns.heatmap(target_corr, annot=True, cmap='coolwarm', fmt='.2f')
```

```python
df['datetime'] = pd.to_datetime(df['datetime'])
df = df.sort_values('datetime')
df.set_index('datetime', inplace=True)

# Choose target and features
features = df.drop(columns=['temp'])  # After dropping correlated ones
target = df['temp']

# Normalize features
scaler_X = StandardScaler()
X_scaled = scaler_X.fit_transform(features)

# Standardize target  
scaler_y = StandardScaler()
y_scaled = scaler_y.fit_transform(target.values.reshape(-1, 1))
```

```python
def create_sequences(X, y, window=7, step=1):
    Xs, ys = [], []
    for i in range(0, len(X) - window, step):
        Xs.append(X[i:i+window])
        ys.append(y[i+window])
    return np.array(Xs), np.array(ys)

window_size = 14  # Increased window size for better context
step_size = 1     # Overlap for more training data
X_seq, y_seq = create_sequences(X_scaled, y_scaled, window=window_size, step=step_size)

split = int(0.8 * len(X_seq))
X_train, X_test = X_seq[:split], X_seq[split:]
y_train, y_test = y_seq[:split], y_seq[split:]


model = Sequential()
model.add(LSTM(64, activation='relu', input_shape=(X_train.shape[1], X_train.shape[2])))
model.add(Dense(1))
model.compile(optimizer='adam', loss='mse')

early_stopping = EarlyStopping(monitor='val_loss', 
                               patience=5,  # Number of epochs with no improvement to wait
                               restore_best_weights=True) 

model.fit(X_train, y_train, epochs=20, batch_size=16, validation_split=0.1, callbacks=[early_stopping])
```

```python
y_pred = model.predict(X_test)

# Inverse transform to original scale
y_pred_inv = scaler_y.inverse_transform(y_pred)
y_test_inv = scaler_y.inverse_transform(y_test)

# Calculate metrics
rmse = np.sqrt(mean_squared_error(y_test_inv, y_pred_inv))
mae = mean_absolute_error(y_test_inv, y_pred_inv)
mape = np.mean(np.abs((y_test_inv - y_pred_inv) / y_test_inv)) * 100

print(f"RMSE: {rmse:.2f}")
print(f"MAE: {mae:.2f}")
print(f"MAPE: {mape:.2f}%")

# Plot predictions with actual values
plt.figure(figsize=(12, 6))
plt.plot(y_test_inv, label='Actual Temperature', color='blue', alpha=0.7)
plt.plot(y_pred_inv, label='Predicted Temperature', color='red', alpha=0.7)
```
